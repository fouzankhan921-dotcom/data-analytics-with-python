### Q4 — Count Frequency of Numbers

```python
numbers = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]

frequency = {}

for num in numbers:
    if num in frequency:
        frequency[num] += 1
    else:
        frequency[num] = 1

print(frequency)
```

### Q5 — Word Counter

### Take sentence input and count word frequency.

```python
sentence = input("Enter a sentence: ")

words = sentence.split()

frequency = {}

for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

print("Word Frequency:")
print(frequency)
```

### Q6 — Student Marks & Average

```python
marks = {"Math": 80, "English": 70, "Physics": 90}

total = 0

for mark in marks.values():
    total += mark

average = total / len(marks)

print("Total:", total)
print("Average:", average)
```

### Q7 — Highest Scoring Student

```python
students = {
    "Ali": 85,
    "Sara": 92,
    "Ahmed": 78,
    "Zara": 95
}

highest_student = ""
highest_marks = 0

for name, marks in students.items():
    if marks > highest_marks:
        highest_marks = marks
        highest_student = name

print("Highest Scoring Student:", highest_student)
print("Marks:", highest_marks)
```

### Q8 — Merge Dictionaries

```python
dict1 = {"a": 1, "b": 2}
dict2 = {"c": 3, "d": 4}

merged = {**dict1, **dict2}

print(merged)
```

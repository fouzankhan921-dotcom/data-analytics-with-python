1. Create a 1D NumPy array of integers from 10 to 50 (inclusive, step 5). Print the array, its shape, dtype and number of dimensions.

```python
# Create a 1D array from 10 to 50 inclusive, with step 5
arr = np.arange(10, 51, 5)

# Print the array and its properties
print("Array:", arr)
print("Shape:", arr.shape)
print("Data Type:", arr.dtype)
print("Number of Dimensions:", arr.ndim)
```

2. Create a 3x4 array of zeros, a 3x4 array of ones, and a 4x4 identity matrix. Then create an array of 8 evenly spaced numbers between 0 and 1 using linspace.

```python
# 1. Create a 3x4 array of zeros
zeros_array = np.zeros((3, 4))
print("3x4 Array of Zeros:")
print(zeros_array)
print("-" * 30)

# 2. Create a 3x4 array of ones
ones_array = np.ones((3, 4))
print("3x4 Array of Ones:")
print(ones_array)
print("-" * 30)

# 3. Create a 4x4 identity matrix
identity_matrix = np.eye(4)
print("4x4 Identity Matrix:")
print(identity_matrix)
print("-" * 30)

# 4. Create an array of 8 evenly spaced numbers between 0 and 1
linspace_array = np.linspace(0, 1, 8)
print("8 Evenly Spaced Numbers between 0 and 1:")
print(linspace_array)
```

3. Create a 1D array of numbers from 1 to 24 and reshape it into a 2D array of shape (4, 6). Then reshape it into a 3D array of shape (2, 3, 4) and print the shape of each.

```python
# 1. Create a 1D array of numbers from 1 to 24
# Note: np.arange's stop value is exclusive, so we use 25 to include 24.

array_1d = np.arange(1, 25)
print("1D Array Shape:", array_1d.shape)
print(array_1d)
print("-" * 20)

# 2. Reshape into a 2D array of shape (4, 6)

array_2d = array_1d.reshape(4, 6)
print("2D Array Shape:", array_2d.shape)
print(array_2d)
print("-" * 20)

# 3. Reshape into a 3D array of shape (2, 3, 4)

array_3d = array_2d.reshape(2, 3, 4)
print("3D Array Shape:", array_3d.shape)
print(array_3d)
```

4. Given arr = np.array([12, 45, 7, 23, 56, 89, 34, 21]), print: the first 3 elements, the last 3 elements, every alternate element, and the array in reverse order.

```python
arr = np.array([12, 45, 7, 23, 56, 89, 34, 21])

# 1. First 3 elements
first_3 = arr[:3]
print("First 3 elements:", first_3)

# 2. Last 3 elements
last_3 = arr[-3:]
print("Last 3 elements:", last_3)

# 3. Every alternate element (starting from the first)
alternate = arr[::2]
print("Every alternate element:", alternate)

# 4. Array in reverse order
reversed_arr = arr[::-1]
print("Reversed array:", reversed_arr)
```

5. Create a 2D array of shape (5, 5) with values 1 to 25. Extract (a) the second row, (b) the third column, (c) the 3x3 sub-matrix from the center.

```python
arr = np.arange(1, 26).reshape(5, 5)
print("Original 5x5 Array:\n", arr)
print("-" * 30)

# (a) Extract the second row
# Index 1 represents the second row
second_row = arr[1, :] 
print("(a) Second row:\n", second_row)
print("-" * 30)

# (b) Extract the third column
# Index 2 represents the third column
third_col = arr[:, 2] 
print("(b) Third column:\n", third_col)
print("-" * 30)

# (c) Extract the 3x3 sub-matrix from the center
# Center starts at index 1 and goes up to (but not including) index 4
center_submatrix = arr[1:4, 1:4] 
print("(c) Center 3x3 sub-matrix:\n", center_submatrix)
```

6. Given a = np.array([10, 20, 30, 40]) and b = np.array([1, 2, 3, 4]), perform addition, subtraction, multiplication, division and power (a ** 2). Also compute the dot product of a and b.

```python
# Define the arrays
a = np.array([10, 20, 30, 40])
b = np.array([1, 2, 3, 4])

print("Array a:", a)
print("Array b:", b)
print("-" * 30)

# Element-wise operations
addition = a + b
subtraction = a - b
multiplication = a * b
division = a / b
power = a ** 2

print("Addition (a + b):        ", addition)
print("Subtraction (a - b):     ", subtraction)
print("Multiplication (a * b):  ", multiplication)
print("Division (a / b):        ", division)
print("Power (a ** 2):          ", power)
print("-" * 30)

# Dot product
dot_product = np.dot(a, b)
print("Dot product (np.dot(a, b)):", dot_product)
```

7. Given arr = np.array([[4, 8, 2], [9, 1, 7], [5, 6, 3]]), find the sum, mean, min, max and standard deviation of the whole array, and then the sum along rows (axis=1) and along columns (axis=0).

```python
# Given 2D array
arr = np.array([[4, 8, 2], 
                [9, 1, 7], 
                [5, 6, 3]])

print("Array:\n", arr)
print("-" * 40)

# Statistics of the whole array
print("Sum of whole array:   ", np.sum(arr))
print("Mean of whole array:  ", np.mean(arr))
print("Min of whole array:   ", np.min(arr))
print("Max of whole array:   ", np.max(arr))
print("Std of whole array:   ", np.std(arr))
print("-" * 40)

# Sum along rows (axis=1)
sum_rows = np.sum(arr, axis=1)
print("Sum along rows (axis=1):   ", sum_rows)

# Sum along columns (axis=0)
sum_cols = np.sum(arr, axis=0)
print("Sum along columns (axis=0):", sum_cols)
```

8. Create an array of 20 random integers between 1 and 100. Using boolean masking, print (a) all elements greater than 50, (b) all even elements, (c) count of elements divisible by 5.

```python
# Set a seed so the random numbers are the same every time you run it (optional)
np.random.seed(42)

# Create an array of 20 random integers between 1 and 100 (inclusive)
# Note: The high parameter in randint is exclusive, so we use 101.
arr = np.random.randint(1, 101, 20)
print("Original Array:")
print(arr)
print("-" * 50)

# (a) All elements greater than 50
greater_than_50 = arr[arr > 50]
print("(a) Elements greater than 50:")
print(greater_than_50)
print("-" * 50)

# (b) All even elements
even_elements = arr[arr % 2 == 0]
print("(b) Even elements:")
print(even_elements)
print("-" * 50)

# (c) Count of elements divisible by 5
# arr % 5 == 0 creates a boolean array (True where divisible by 5). 
# np.sum() adds up the True values (True = 1, False = 0).
count_divisible_by_5 = np.sum(arr % 5 == 0)
print("(c) Count of elements divisible by 5:", count_divisible_by_5)
```

9. Use np.where to replace all negative values in np.array([5, -3, 8, -1, 0, -7, 4]) with 0. Also find the index of the maximum and minimum value using argmax and argmin.

```python
# Given array
arr = np.array([5, -3, 8, -1, 0, -7, 4])
print("Original Array:", arr)
print("-" * 40)

# 1. Replace all negative values with 0 using np.where
# np.where(condition, value_if_true, value_if_false)
arr_replaced = np.where(arr < 0, 0, arr)
print("Array after replacing negatives with 0:")
print(arr_replaced)
print("-" * 40)

# 2. Find the index of the maximum value
max_index = np.argmax(arr)
print(f"Index of Maximum value ({arr[max_index]}): {max_index}")

# 3. Find the index of the minimum value
min_index = np.argmin(arr)
print(f"Index of Minimum value ({arr[min_index]}): {min_index}")
```

10. Create two 2x3 arrays and join them vertically (vstack) and horizontally (hstack). Then sort the array np.array([[9, 3, 6], [2, 8, 1]]) along each row and flatten the result into 1D.

```python
# 1. Create two 2x3 arrays
a = np.array([[1, 2, 3], 
              [4, 5, 6]])

b = np.array([[7, 8, 9], 
              [10, 11, 12]])

print("Array a:\n", a)
print("Array b:\n", b)
print("-" * 40)

# 2. Join them vertically (vstack)
v_stack = np.vstack((a, b))
print("Vertical Stack (vstack) - Shape:", v_stack.shape)
print(v_stack)
print("-" * 40)

# 3. Join them horizontally (hstack)
h_stack = np.hstack((a, b))
print("Horizontal Stack (hstack) - Shape:", h_stack.shape)
print(h_stack)
print("-" * 40)

# 4. Sort the given array along each row and flatten it
arr = np.array([[9, 3, 6], 
                [2, 8, 1]])

# Sort along each row (axis=1 means operate across columns)
sorted_rows = np.sort(arr, axis=1)
print("Sorted along each row:\n", sorted_rows)

# Flatten the result into a 1D array
flattened = sorted_rows.flatten()
print("Flattened 1D array:", flattened)
```

### QUESTION 1: Working with Lists
### 1.Create a list called numbers containing at least 8 integers.
### 2.Using a for loop, do the following:
### Print all numbers in the list.
### Print the sum of all numbers.
### Print only the even numbers.
### Requirements:
### Use at least one loop.
### Use conditional statements.
### Display clear output.

```python
# 1. Create a list with at least 8 integers
numbers = [9, 8, 32, 13, 37, 44, 16, 7]

# 2. Use a for loop and conditional statements
total = 0
even_numbers = []

print("All numbers in the list:")
for num in numbers:
    print(num)
    total += num

    if num % 2 == 0:
        even_numbers.append(num)

print("\nSum of all numbers:", total)

print("\nEven numbers only:")
for num in even_numbers:
    print(num)
```

### QUESTION 2: Dictionary Basics
### 1.Create a dictionary called student with the following keys:
### name
### age
### course
### marks (must store a list of 3 marks)
### 2.Using loops:
### Print all keys and their values.
### Calculate the average of the marks.
### Display whether the student Passed (average ≥ 50) or Failed.
### Requirements:
### Dictionary must contain a list.
### Use loops to calculate the average.

```python
# 1. Create a dictionary called student
student = {
    "name": "Arsalan",
    "age": 26,
    "course": "Data Science",
    "marks": [78, 84, 56]   # list of 3 marks
}

# 2. Print all keys and their values using a loop
print("Student Details:")
for key, value in student.items():
    print(f"{key}: {value}")

# Calculate the average of the marks using a loop
total = 0
for mark in student["marks"]:
    total += mark

average = total / len(student["marks"])

print(f"\nAverage marks: {average:.2f}")

# Display Passed or Failed
if average >= 50:
    print("Result: Passed")
else:
    print("Result: Failed")
```

### QUESTION 3: List of Dictionaries
### 1.Create a list called employees.
### 2.Each employee must be stored as a dictionary containing:
### name
### department
### salary
### 3.Using loops:
### Print details of all employees.
### Find the employee with the highest salary.
### Calculate the total salary expense.
### Requirements:
### Minimum 3 employees.
### Use comparison logic.
### Use loops to process data.

```python
# 1. Create a list called employees
employees = [
    {"name": "Ilyas",   "department": "Electrical",      "salary": 58000},
    {"name": "Ahmed",     "department": "HR",      "salary": 69000},
    {"name": "Qadir", "department": "Finance", "salary": 76000},
    {"name": "Jawad",   "department": "IT",      "salary": 83000}
]

# 2. Print details of all employees using a loop
print("Employee Details:")
print("-" * 40)
for emp in employees:
    print(f"Name: {emp['name']}")
    print(f"Department: {emp['department']}")
    print(f"Salary: {emp['salary']}")
    print("-" * 40)

# 3. Find the employee with the highest salary using comparison logic
highest_paid = employees[0]   # start with the first employee
for emp in employees:
    if emp["salary"] > highest_paid["salary"]:
        highest_paid = emp

print(f"\nHighest Paid Employee: {highest_paid['name']} "
      f"({highest_paid['department']}) - {highest_paid['salary']}")

# 4. Calculate the total salary expense using a loop
total_salary = 0
for emp in employees:
    total_salary += emp["salary"]

print(f"\nTotal Salary Expense: {total_salary}")
```

### QUESTION 4: While Loop with User Input
### 1.Ask the user to enter numbers.
### 2.Store the numbers in a list.
### 3.Stop when the user enters -1.
### 4.After the loop ends:
### Print the complete list.
### Print the largest number.
### Print the smallest number.
### Requirements:
### Must use a while loop.
### Must store input in a list.
### Handle user input correctly.

```python
# 1. Create an empty list to store numbers
numbers = []

# 2. Ask the user to enter numbers using a while loop
print("Enter numbers one by one.")
print("Type -1 to stop.\n")

while True:
    user_input = input("Enter a number (or -1 to stop): ")

    # Handle non-numeric input gracefully
    try:
        num = int(user_input)
    except ValueError:
        print("Invalid input! Please enter a valid integer.\n")
        continue

    # Stop the loop when the user enters -1
    if num == -1:
        break

    # Store the number in the list
    numbers.append(num)

# 3. After the loop ends, display results
print("\n" + "=" * 40)

if len(numbers) == 0:
    print("No numbers were entered (the list is empty).")
else:
    # Print the complete list
    print(f"Complete list: {numbers}")

    # Find largest and smallest using comparison logic
    largest = numbers[0]
    smallest = numbers[0]

    for n in numbers:
        if n > largest:
            largest = n
        if n < smallest:
            smallest = n

    print(f"Largest number : {largest}")
    print(f"Smallest number: {smallest}")

print("=" * 40)
```

### QUESTION 5: Word Frequency Counter
### 1.Ask the user to enter a sentence.
### 2.Convert the sentence into a list of words.
### 3.Use a dictionary to count how many times each word appears.
### 4.Print the word frequency clearly.
### Example Format (for understanding only):
### apple: 3
### banana: 2
### orange: 1
### Requirements:
### Use string splitting.
### Use a dictionary for counting.
### Use loops for processing.

```python
# 1. Ask the user to enter a sentence
sentence = input("Enter a sentence: ")

# 2. Convert the sentence into a list of words (string splitting)
#    .lower() makes counting case-insensitive ("The" and "the" count as one word)
words = sentence.lower().split()

# 3. Use a dictionary to count how many times each word appears
word_count = {}

for word in words:
    if word in word_count:
        word_count[word] += 1   # word already seen → increase count
    else:
        word_count[word] = 1    # first time seeing this word

# 4. Print the word frequency clearly
print("\nWord Frequency Counter:")
print("-" * 30)

for word, count in word_count.items():
    print(f"{word}: {count}")
```

### QUESTION 6: Nested Loops – Multiplication Table
### Using nested loops, print a multiplication table from 1 to 5.
### Expected Output Format:
### 1 2 3 4 5
### 2 4 6 8 10
### 3 6 9 12 15
### 4 8 12 16 20
### 5 10 15 20 25
### Requirements:
### Must use nested loops.
### Proper formatting required.

```python
# Nested loops to print a 5x5 multiplication table
for i in range(1, 5):          # outer loop → rows (1 to 5)
    for j in range(1, 5):      # inner loop → columns (1 to 5)
        print(i * j, end=" ")  # print product on the same line
    print()                    # move to the next line after each row
```

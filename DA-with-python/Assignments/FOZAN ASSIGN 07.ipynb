2. Create a Series of 5 numbers with custom index labels ['a','b','c','d','e']. Print the value at label 'c', the values at positions 1 to 3, and the sum and mean of the Series.


```python
# 1. Create a Series of 5 numbers with custom index labels
data = [10, 20, 30, 40, 50]
labels = ['a', 'b', 'c', 'd', 'e']

s = pd.Series(data, index=labels)
print("Original Series:")
print(s)
print("-" * 30)

# 2. Print the value at label 'c'
# We use .loc[] for label-based indexing
value_c = s.loc['c']
print("Value at label 'c':", value_c)
print("-" * 30)

# 3. Print the values at positions 1 to 3
# We use .iloc[] for position-based indexing. 
# Slicing 1:4 gives positions 1, 2, and 3 (the end index is exclusive).
values_1_to_3 = s.iloc[1:4]
print("Values at positions 1 to 3:")
print(values_1_to_3)
print("-" * 30)

# 4. Print the sum and mean of the Series
print("Sum of Series: ", s.sum())
print("Mean of Series:", s.mean())
```

4. From the DataFrame of Q1, select (a) only the Name and Salary columns, (b) rows where Age is greater than 30, (c) rows where City is 'Delhi' and Salary is above 40000.

```python
# Recreating the DataFrame from Q1
data = {
    'Name': ['Aryaan', 'Babar', 'Mahad', 'Danish', 'Manahil', 'Falak', 'Grover', 'Hania'],
    'Age': [25, 30, 35, 28, 42, 31, 29, 38],
    'City': ['New York', 'London', 'Paris', 'Tokyo', 'Berlin', 'Sydney', 'Toronto', 'Madrid'],
    'Salary': [55000, 62000, 75000, 58000, 90000, 68000, 72000, 81000]
}
df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)
print("=" * 50)

# (a) Select only the Name and Salary columns
# We pass a list of column names inside double square brackets
print("(a) Only 'Name' and 'Salary' columns:")
print(df[['Name', 'Salary']])
print("=" * 50)

# (b) Select rows where Age is greater than 30
# We use a boolean condition inside the square brackets
print("(b) Rows where Age > 30:")
print(df[df['Age'] > 30])
print("=" * 50)

# (c) Select rows where City is 'Delhi' and Salary is above 40000
# We combine two conditions using the '&' operator.
# Note: Each condition MUST be wrapped in parentheses.
print("(c) Rows where City is 'Delhi' and Salary > 40000:")
print(df[(df['City'] == 'Delhi') & (df['Salary'] > 40000)])
```

5. Use loc and iloc to (a) get the 2nd row, (b) get Name of the 4th employee, (c) get rows 1 to 3 with columns Name and City.


```python
data = {
    'Name': ['Aryaan', 'Babar', 'Mahad', 'Danish', 'Manahil', 'Falak', 'Grover', 'Hania'],
    'Age': [25, 30, 35, 28, 42, 31, 29, 38],
    'City': ['New York', 'London', 'Paris', 'Tokyo', 'Berlin', 'Sydney', 'Toronto', 'Madrid'],
    'Salary': [55000, 62000, 75000, 58000, 90000, 68000, 72000, 81000]
}
df = pd.DataFrame(data)

# Using loc (label-based) - The 2nd row has index label 1
row_loc = df.loc[1]

# Using iloc (position-based) - The 2nd row is at integer position 1
row_iloc = df.iloc[1]

print("(a) 2nd row using loc:\n", row_loc)
print("\n(a) 2nd row using iloc:\n", row_iloc)

# Using loc - 4th employee has index label 3, column name 'Name'
name_loc = df.loc[3, 'Name']

# Using iloc - 4th employee is at position 3, 'Name' is column position 0
name_iloc = df.iloc[3, 0]

print("(b) Name of 4th employee using loc: ", name_loc)
print("(b) Name of 4th employee using iloc:", name_iloc)

# Using loc - loc includes the endpoint, so rows 1, 2, 3
subset_loc = df.loc[1:3, ['Name', 'City']]

# Using iloc - iloc excludes the endpoint, so we use 1:4 for positions 1, 2, 3.
# 'Name' is position 0, 'City' is position 2.
subset_iloc = df.iloc[1:4, [0, 2]]

print("(c) Rows 1-3 with Name & City using loc:\n", subset_loc)
print("\n(c) Rows 1-3 with Name & City using iloc:\n", subset_iloc)
```

6. Add a new column 'Bonus' that is 10% of Salary and a column 'Total' = Salary + Bonus. Then rename the column 'City' to 'Location' and drop the 'Bonus' column.

```python
data = {
    'Name': ['Aryaan', 'Babar', 'Mahad', 'Danish', 'Manahil', 'Falak', 'Grover', 'Hania'],
    'Age': [25, 30, 35, 28, 42, 31, 29, 38],
    'City': ['New York', 'London', 'Paris', 'Tokyo', 'Berlin', 'Sydney', 'Toronto', 'Madrid'],
    'Salary': [55000, 62000, 75000, 58000, 90000, 68000, 72000, 81000]
}
df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)
print("=" * 50)

# 1. Add a new column 'Bonus' that is 10% of Salary
df['Bonus'] = df['Salary'] * 0.10
print("After adding 'Bonus' column:")
print(df)
print("=" * 50)

# 2. Add a column 'Total' = Salary + Bonus
df['Total'] = df['Salary'] + df['Bonus']
print("After adding 'Total' column:")
print(df)
print("=" * 50)

# 3. Rename the column 'City' to 'Location'
# We pass a dictionary mapping old names to new names and use inplace=True to modify df directly.
df.rename(columns={'City': 'Location'}, inplace=True)
print("After renaming 'City' to 'Location':")
print(df)
print("=" * 50)

# 4. Drop the 'Bonus' column
# We specify columns=['Bonus'] and use inplace=True to modify df directly.
df.drop(columns=['Bonus'], inplace=True)
print("Final DataFrame after dropping 'Bonus':")
print(df)
```

7. Create a DataFrame having some missing values (NaN). Find the count of missing values in each column, fill numeric NaNs with the column mean, and drop rows that still have missing values.

```python
import numpy as np
data = {
    'Name': ['Aryaan', 'Babar', np.nan , 'Danish', 'Manahil', 'Falak', 'Grover', 'Hania'],
    'Age': [25, np.nan , 35, 28, 42, 31, 29, 38],
    'City': ['New York', 'London', 'Paris', np.nan , 'Berlin', 'Sydney', 'Toronto', 'Madrid'],
    'Salary': [55000, np.nan , 75000, 58000, 90000, 68000, np.nan , 81000]
}
df = pd.DataFrame(data)

print("Original DataFrame with Missing Values:")
print(df)
print("=" * 50)

# 2. Count missing values in each column
print("Missing values count per column:")
print(df.isnull().sum())
print("=" * 50)

# 3. Fill numeric NaNs with the column mean
# We select only numeric columns using select_dtypes, then fillna with mean.
numeric_cols = df.select_dtypes(include=[np.number]).columns
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].mean())

print("After filling numeric NaNs with column mean:")
print(df)
print("=" * 50)

# 4. Drop rows that still have any missing values (e.g., NaN in Name or City)
df_cleaned = df.dropna()

print("After dropping rows with remaining NaNs:")
print(df_cleaned)
```

8. Sort the DataFrame by Salary in descending order, and then by Age in ascending order. Also print the top 3 highest-paid employees using nlargest().

```python
data = {
    'Name': ['Aryaan', 'Babar', 'Mahad', 'Danish', 'Manahil', 'Falak', 'Grover', 'Hania'],
    'Age': [25, 30, 35, 28, 42, 31, 29, 38],
    'City': ['New York', 'London', 'Paris', 'Tokyo', 'Berlin', 'Sydney', 'Toronto', 'Madrid'],
    'Salary': [55000, 62000, 75000, 58000, 90000, 68000, 72000, 81000]
}
df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)
print("=" * 50)

# 1. Sort by Salary descending, then by Age ascending
sorted_df = df.sort_values(by=['Salary', 'Age'], ascending=[False, True])

print("DataFrame sorted by Salary (desc) and Age (asc):")
print(sorted_df)
print("=" * 50)

# 2. Top 3 highest-paid employees using nlargest()
top_3 = df.nlargest(3, 'Salary')

print("Top 3 highest-paid employees:")
print(top_3)
```

9. Using groupby, find the average salary per City, the number of employees per City, and the maximum salary per City. Show the result using agg() with multiple functions.

```python
data = {
    'Name': ['Aryaan', 'Babar', 'Mahad', 'Danish', 'Manahil', 'Falak'],
    'Age': [25, 30, 35, 28, 42, 31],
    'City': ['New York', 'London', 'Paris', 'Tokyo', 'Berlin', 'Sydney'],
    'Salary': [55000, 62000, 75000, 58000, 90000, 68000]
}
df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)
print("=" * 60)

# 1. Average salary per City
avg_salary = df.groupby('City')['Salary'].mean()
print("Average Salary per City:")
print(avg_salary)
print("=" * 60)

# 2. Number of employees per City
emp_count = df.groupby('City')['Name'].count()
print("Number of Employees per City:")
print(emp_count)
print("=" * 60)

# 3. Maximum salary per City
max_salary = df.groupby('City')['Salary'].max()
print("Maximum Salary per City:")
print(max_salary)
print("=" * 60)

# 4. Using agg() with multiple functions at once
result = df.groupby('City')['Salary'].agg(['mean', 'count', 'max'])
print("Using agg() with multiple functions:")
print(result)
print("=" * 60)

# Bonus: Different functions applied to different columns at once
result_multi = df.groupby('City').agg(
    Avg_Salary=('Salary', 'mean'),
    Emp_Count=('Name', 'count'),
    Max_Salary=('Salary', 'max'),
    Avg_Age=('Age', 'mean')
)
print("Multiple aggregations with custom column names:")
print(result_multi)
```

10. Create two DataFrames (employees and departments) with a common column 'DeptID'. Merge them using inner and left joins, then concatenate two DataFrames of the same structure using pd.concat() and remove duplicate rows.

```python
# 1. Create Employees DataFrame
employees = pd.DataFrame({
    'EmpID': [101, 102, 103, 104],
    'Name': ['Aryaan', 'Babar', 'Mahad', 'Danish'],
    'DeptID': [1, 2, 3, 2]
})

# 2. Create Departments DataFrame
departments = pd.DataFrame({
    'DeptID': [1, 2, 4],
    'Department': ['HR', 'IT', 'Finance']
})

print("Employees DataFrame:")
print(employees)

print("\nDepartments DataFrame:")
print(departments)

# 3. Inner Join
inner_merge = pd.merge(employees, departments, on='DeptID', how='inner')

print("\nInner Join:")
print(inner_merge)

# 4. Left Join
left_merge = pd.merge(employees, departments, on='DeptID', how='left')

print("\nLeft Join:")
print(left_merge)

# 5. Create two DataFrames of the same structure
df1 = pd.DataFrame({
    'EmpID': [101, 102, 103],
    'Name': ['Aryaan', 'Babar', 'Mahad']
})

df2 = pd.DataFrame({
    'EmpID': [102, 103, 105],
    'Name': ['Aryan', 'Mahad', 'Manahil']
})

# 6. Concatenate DataFrames
combined = pd.concat([df1, df2], ignore_index=True)

print("\nAfter Concatenation:")
print(combined)

# 7. Remove duplicate rows
unique_rows = combined.drop_duplicates()

print("\nAfter Removing Duplicates:")
print(unique_rows)
```
